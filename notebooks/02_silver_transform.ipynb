```python
df_bronze = spark.table("bronze_jobs")

print("Bronze records: ", df_bronze.count())
display(df_bronze.limit(10))
```

```python
from pyspark.sql import functions as F

salary_pattern = r"(\d+(?:\.\d+)?)\s*-\s*(\d+(?:\.\d+)?)\s*triệu"

df_silver = (
    df_bronze

    .withColumn(
        "_salary_min_raw",
        F.regexp_extract(
            F.lower(F.col("salary_raw")),
            salary_pattern,
            1
        )
    )

    .withColumn(
        "_salary_max_raw",
        F.regexp_extract(
            F.lower(F.col("salary_raw")),
            salary_pattern,
            2
        )
    )
)
```

```python
df_silver = (
    df_silver

    .withColumn(
        "salary_min",
        F.when(
            F.col("_salary_min_raw") != "",
            F.col("_salary_min_raw").cast("double") * 1_000_000
        )
    )

    .withColumn(
        "salary_max",
        F.when(
            F.col("_salary_max_raw") != "",
            F.col("_salary_max_raw").cast("double") * 1_000_000
        )
    )
)
```

```python
df_silver = df_silver.withColumn(
    "salary_type",

    F.when(
        F.lower(F.col("salary_raw")).contains("thoả thuận"),
        "negotiable"
    )

    .when(
        F.col("salary_min").isNotNull()
        & F.col("salary_max").isNotNull(),
        "range"
    )

    .otherwise("unsupported")
)
```

```python
df_silver = df_silver.withColumn(
    "currency",

    F.when(
        F.col("salary_type") == "range",
        F.lit("VND")
    )
)
```

```python
df_silver = df_silver.withColumn(
    "city_raw",
    F.trim(
        F.split(
            F.col("job_location_raw"),
            ":"
        ).getItem(0)
    )
)
```

```python
df_silver = df_silver.withColumn(
    "address",
    F.trim(
        F.regexp_extract(
            F.col("job_location_raw"),
            r"^[^:]+:(.*)$",
            1
        )
    )
)
```

```python
df_silver = df_silver.withColumn(
    "city",

    F.when(
        F.col("city_raw").isin(
            "TP.HCM",
            "TP. HCM",
            "TPHCM"
        ),
        "Hồ Chí Minh"
    )

    .when(
        F.col("city_raw") == "Hà Nội",
        "Hà Nội"
    )

    .otherwise(F.col("city_raw"))
)
```

```python
title = F.lower(F.col("title"))

it_pattern = (
    r"developer|lập trình|software|phần mềm|"
    r"tester|testing|qa|qc|kiểm thử|"
    r"business analyst|it ba|"
    r"data analyst|data engineer|phân tích dữ liệu|"
    r"\bai\b|machine learning|"
    r"devops|cloud|azure|aws|"
    r"database|dba|"
    r"helpdesk|it support|"
    r"network|infrastructure|"
    r"full.?stack|backend|frontend|"
    r"ui\s*/?\s*ux|"
    r"it project|"
    r"cntt|công nghệ thông tin|"
    r"erp|odoo|"
    r"website|web developer"
)

df_silver = df_silver.withColumn(
    "is_it_job",
    title.rlike(it_pattern)
)
```

```python

df_silver = df_silver.withColumn(
    "job_title_group",

    F.when(~F.col("is_it_job"), "Non-IT")

    .when(
        title.rlike(r"tester|testing|qa|qc|kiểm thử"),
        "QA / Tester"
    )

    .when(
        title.rlike(r"business analyst|it ba|phân tích nghiệp vụ"),
        "Business Analyst"
    )

    .when(
        title.rlike(r"data analyst|data engineer|phân tích dữ liệu"),
        "Data"
    )

    .when(
        title.rlike(r"\bai\b|machine learning"),
        "AI / Machine Learning"
    )

    .when(
        title.rlike(r"devops|cloud|azure|aws"),
        "DevOps / Cloud"
    )

    .when(
        title.rlike(r"database|\bdba\b"),
        "Database"
    )

    .when(
        title.rlike(r"helpdesk|it support|technical support"),
        "IT Support / Helpdesk"
    )

    .when(
        title.rlike(r"network|infrastructure"),
        "Network / Infrastructure"
    )

    .when(
        title.rlike(r"full.?stack"),
        "Fullstack Developer"
    )

    .when(
        title.rlike(r"backend"),
        "Backend Developer"
    )

    .when(
        title.rlike(r"frontend"),
        "Frontend Developer"
    )

    .when(
        title.rlike(r"developer|lập trình|software engineer|kỹ sư.*phần mềm"),
        "Software Developer"
    )

    .when(
        title.rlike(r"ui\s*/?\s*ux|designer"),
        "UI / UX"
    )

    .when(
        title.rlike(r"project manager|quản lý dự án"),
        "Project Management"
    )

    .when(
        title.rlike(r"product manager|product management"),
        "Product Management"
    )

    .when(
        title.rlike(r"triển khai.*phần mềm|triển khai.*erp"),
        "Software Implementation"
    )

    .otherwise("Other IT")
)
```

```python
df_silver = df_silver.withColumn(
    "job_level",

    F.when(~F.col("is_it_job"), "Not Applicable")

    .when(
        title.rlike(r"intern|internship|thực tập"),
        "Intern"
    )

    .when(
        title.rlike(r"fresher"),
        "Fresher"
    )

    .when(
        title.rlike(r"junior"),
        "Junior"
    )

    .when(
        title.rlike(r"senior"),
        "Senior"
    )

    .when(
        title.rlike(r"leader|lead|trưởng nhóm|trưởng phòng"),
        "Lead / Manager"
    )

    .otherwise("Unspecified")
)
```

```python
(
    df_silver.write
    .option("mergeSchema", "true")
    .format("delta")
    .mode("overwrite")
    .saveAsTable("silver_jobs")
)
```
